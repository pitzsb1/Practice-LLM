```python
import pandas as pd

meeting_note_csv_path = "./guitar_final.csv"
df_rttm = pd.read_csv(meeting_note_csv_path, sep='|')

display(df_rttm)
```

```python
# 이름 넣기
name_dict = {
    "SPEAKER_00": "AI",
    "SPEAKER_01": "이성용",
}

df_rttm["name"] = df_rttm["speaker_id"].apply(lambda x: name_dict[x])

display(df_rttm)
```

```python
meeting_note_txt = df_rttm[['start', 'end', 'name', 'text']].to_json(orient='records', force_ascii=False)
print(meeting_note_txt)
```

```python
system_prompt = f'''
너는 회의 내용을 요약하는 봇이다. 아래 회의록을 읽고 회의록 양식에 맞게 정리하라.
반드시 아래 키를 가진 JSON 객체로만 답하라. 회의록에서 알 수 없는 항목은 빈 문자열("")로 둔다.

- date: 날짜
- time: 시간
- place: 장소
- writer: 작성자
- attendees: 참석자 (문자열)
- purpose: 목적(목표) (문자열)
- topic: 주제 (문자열)
- content: 회의 내용 (문자열 리스트, 핵심 내용을 항목별로)
- conclusion: 결론 및 합의 (문자열 리스트)
- todo: 할 일 (문자열 리스트)
- notes: 참고 사항 (문자열)

=============== 이하 회의록 ===============

{ meeting_note_txt }
'''
```

```python
from openai import OpenAI
from dotenv import load_dotenv
import os

load_dotenv()
api_key = os.getenv('OPEN_API_KEY')
client = OpenAI(api_key=api_key)
```

```python
import json

response = client.chat.completions.create(
    model='gpt-4o',
    temperature=0.1,
    response_format={"type": "json_object"},
    messages=[{"role": "system", "content": system_prompt}]
)

summary = response.choices[0].message.content.strip()
summary_json = json.loads(summary)
print(summary_json)
```

```python
with open('./guitar_summary.md', 'w', encoding='utf-8') as f:
    f.write(summary)
```

```python
df_meeting_note = df_rttm[['start', 'end', 'name', 'text']].copy()
df_meeting_note.dropna(inplace=True)

meeting_note_dict = df_meeting_note.to_dict(orient='records')  # 딕셔너리 형태로 변환
meeting_note_dict
```

```python
for row in meeting_note_dict:    
    if not isinstance(row['text'], str): # 텍스트가 아닌 다른 형식인 경우 통과
        continue

    correction_system_prompt = f'''
        너는 주어진 회의 녹취록을 수정 및 보완하는 봇이다.
        주어진 회의 녹취록은 STT로 작성된 결과이므로, 이 중에 오류가 있는 부분을 찾아내고 수정하라.

        표현, 말투는 최대한 원본과 일치하도록 유지하되, 잘못 받아적은 내용을 수정하라.
        원본의 내용을 최대한 살리되, 잘못된 내용만 수정하라. 
        어떤 경우에는 A라는 사람이 말한 내용을 B가 말한 것으로 잘못 기록된 경우도 있을 수 있다. 이런 경우에는 문맥을 고려하여 수정하라.

        원본 내용 중 빠지거나, 추가되거나, 잘못 기록된 부분을 찾아내어 수정하라. 수정된 메시지 이외에는 아무것도 작성하지 말라.
        --------------------------------
        회의 요약문 : {summary}
        --------------------------------
        회의 녹취록 전문: {df_meeting_note.to_json(orient='records', force_ascii=False)}
        --------------------------------
        확인할 메시지 원본: {row['text']}
    '''

    # limit token 을 초과했을 때 발생하는 에러는 무시하고 다음 Cell 진행
    response = client.chat.completions.create(
        model='gpt-4o',
        temperature=0.1,
        messages=[
            {"role": "system", "content": correction_system_prompt},
        ]
    )

    correction = response.choices[0].message.content
    if ':' in correction:  # “수정: “ 과 같은 형식으로 반환되는 경우 해결 
        correction = correction.split(':')[-1]
        
    correction = correction.strip() # 좌우 공백 제거

    print(row['name'])
    print(row['text'])
    print(correction)
    print('-----------------')

    row['corrected_text'] = correction
```

```python
# meeting_note_dict json 파일로 저장
import json

with open('./guitar_meeting_note_corrected.json', 'w', encoding='utf-8') as f:
    json.dump(meeting_note_dict, f, ensure_ascii=False, indent=4)
```

```python
md_template = ""

for row in meeting_note_dict:
    # corrected_text 키가 있는 경우 (Limit Token 문제로 인해 없는 경우가 있음)
    if 'corrected_text' in row:
        md_template += f"- **{row['name']}** : {row['corrected_text']} \n"
    else:
        md_template += f"- **{row['name']}** : {row['text']} \n"
with open('./guitar_meeting_note_corrected.md', 'w', encoding='utf-8') as f:
    f.write(md_template)
```

```python
from docx import Document
from docx.shared import Pt, Cm
from docx.enum.text import WD_ALIGN_PARAGRAPH
from docx.enum.table import WD_TABLE_ALIGNMENT, WD_ALIGN_VERTICAL
from docx.oxml import OxmlElement
from docx.oxml.ns import qn

def to_text(v):  # 리스트면 "- 항목" 줄바꿈 형태로
    return "\n".join(f"- {x}" for x in v) if isinstance(v, list) else (v or "")

def shade(cell, color="DDE9D6"):  # 라벨 칸 연두색 배경
    tcPr = cell._tc.get_or_add_tcPr()
    shd = OxmlElement('w:shd')
    shd.set(qn('w:val'), 'clear'); shd.set(qn('w:color'), 'auto'); shd.set(qn('w:fill'), color)
    tcPr.append(shd)

def set_cell(cell, text, bold=False, center=False):
    cell.text = ""
    for i, line in enumerate(text.split("\n") if text else [""]):
        p = cell.paragraphs[0] if i == 0 else cell.add_paragraph()
        run = p.add_run(line); run.bold = bold; run.font.size = Pt(10.5)
        run.font.name = "맑은 고딕"
        run._element.rPr.rFonts.set(qn('w:eastAsia'), "맑은 고딕")
        if center: p.alignment = WD_ALIGN_PARAGRAPH.CENTER
    cell.vertical_alignment = WD_ALIGN_VERTICAL.CENTER if center else WD_ALIGN_VERTICAL.TOP

document = Document()
document.add_heading('회의록', level=1).alignment = WD_ALIGN_PARAGRAPH.CENTER

table = document.add_table(rows=0, cols=4)
table.style = 'Table Grid'
table.alignment = WD_TABLE_ALIGNMENT.CENTER
widths = [Cm(2.8), Cm(5.2), Cm(2.8), Cm(5.2)]

def add_row(height):
    row = table.add_row()
    for c, w in zip(row.cells, widths): c.width = w
    row.height = Cm(height)
    return row

# 날짜 | 시간  /  장소 | 작성자
for l1, k1, l2, k2 in [("날짜", "date", "시간", "time"), ("장소", "place", "작성자", "writer")]:
    r = add_row(1.0).cells
    for i, (txt, is_label) in enumerate([(l1, True), (to_text(summary_json[k1]), False),
                                         (l2, True), (to_text(summary_json[k2]), False)]):
        set_cell(r[i], txt, bold=is_label, center=is_label)
        if is_label: shade(r[i])

# 나머지 항목: 라벨 + 오른쪽 칸 병합
rows = [("참석자", "attendees", 1.0), ("목적(목표)", "purpose", 1.0), ("주제", "topic", 1.0),
        ("회의 내용", "content", 8.0), ("결론 및 합의", "conclusion", 2.5),
        ("할 일", "todo", 2.5), ("참고 사항", "notes", 1.5)]
for label, key, h in rows:
    r = add_row(h).cells
    set_cell(r[0], label, bold=True, center=True); shade(r[0])
    set_cell(r[1].merge(r[3]), to_text(summary_json.get(key, "")))

document.save('./guitar_summary.docx')
```
